```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


# Load Data
df = pd.read_csv('../data/iris.csv', encoding='latin-1')
# Use info function to examine feature names
df.info()
```

```python
# Drop any samples of the Virginica type
# df['iris_type'].unique()
df = df[df['iris_type'] != 'Iris-virginica']
df['iris_type'].unique()
```

```python
# Is there a class imbalance?
df['iris_type'].value_counts().plot(kind='bar')
plt.show()
# No
```
